### STEP 03. 채용공고 페이지 접근 테스트

### 작업 계획

이번 단계에서는 실제 데이터 수집 전에
채용공고 검색 페이지에 HTTP 요청이 가능한지 확인한다.

확인할 항목:
- HTTP 상태 코드
- 응답 Content-Type
- 응답 길이
- HTML 문서가 정상적으로 반환되는지

아직 여러 페이지를 반복 수집하지 않는다.

```python
import requests

url = "https://www.jobkorea.co.kr/Search/?stext=AI"

response = requests.get(url, timeout=10)

print("상태 코드:", response.status_code)
print("Content-Type:", response.headers.get("Content-Type"))
print("응답 길이:", len(response.text))
print("HTML 시작 부분:")
print(response.text[:500])
```

### 실행 결과 해석 / 분석 / 요약

- HTTP 요청 자체는 상태 코드 200으로 성공했다.
- Content-Type도 text/html로 확인되었다.
- 하지만 반환된 HTML은 실제 채용공고 검색 결과가 아니라 보안정책 페이지였다.
- 따라서 현재 requests 방식으로는 원하는 채용공고 데이터를 정상적으로 가져올 수 없다고 판단했다.
- 상태 코드 200만으로 정상적인 데이터 수집이라고 판단하면 안 된다는 점을 확인했다.
- 다음 단계에서는 사이트 제한을 우회하기보다 샘플 HTML 또는 CSV 사용 여부를 결정해야 한다.
