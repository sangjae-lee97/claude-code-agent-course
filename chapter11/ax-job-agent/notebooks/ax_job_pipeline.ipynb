### STEP 06. 전처리 / 중복 제거

### 작업 계획

STEP 05에서 `df_jobs`의 기본 구조에 이상이 없음을 확인했다.
이번 단계에서는 이후 분석에 사용할 수 있도록 데이터를 정리(전처리)한다.

할 일:
- 전처리 전 `df_jobs`의 크기(shape)와 데이터 타입(dtypes) 확인
- 날짜 컬럼(`posted_date`, `closing_date`, `collected_at`)을 datetime 타입으로 변환
  - `pd.to_datetime(..., errors="coerce")` 사용 → 날짜로 바꿀 수 없는 값은 `NaT`(날짜 결측값)가 된다
- 문자열 컬럼(`company_name`, `job_title`, `career`, `location`, `job_url`, `search_keyword`)의 앞뒤 공백 제거
- `job_url` 기준 중복 제거
- 전처리 후 shape / dtypes / 결측값 / `job_url` 중복 개수 재확인

이번 단계에서 하지 않는 것:
- STEP 07 신규 공고 판별
- 실제 크롤링 재시도, Gemini API, Slack / Gmail

완료 조건:
- 날짜 컬럼 3개의 dtype이 datetime으로 바뀐다
- 날짜 변환 후 결측값(`NaT`)이 새로 생기지 않는다
- `job_url` 중복 개수가 0이다


````python
# ----- 1) 전처리 전 상태 확인 -----
shape_before = df_jobs.shape
print("[전처리 전] 데이터 크기:", shape_before)
print("\n[전처리 전] 데이터 타입:")
display(df_jobs.dtypes)

# ----- 2) 날짜 컬럼을 datetime 타입으로 변환 -----
date_columns = ["posted_date", "closing_date", "collected_at"]
for col in date_columns:
    # errors="coerce": 날짜로 바꿀 수 없는 값은 에러를 내지 않고 NaT(날짜 결측값)로 바꾼다
    df_jobs[col] = pd.to_datetime(df_jobs[col], errors="coerce")

# ----- 3) 문자열 컬럼의 앞뒤 공백 제거 -----
text_columns = ["company_name", "job_title", "career", "location", "job_url", "search_keyword"]
for col in text_columns:
    # .str.strip(): " 회사A " → "회사A" 처럼 앞뒤 공백만 제거한다
    df_jobs[col] = df_jobs[col].str.strip()

# ----- 4) job_url 기준 중복 제거 -----
# 같은 job_url이 여러 번 있으면 첫 번째 행만 남긴다
df_jobs = df_jobs.drop_duplicates(subset=["job_url"])

# ----- 5) 전처리 후 상태 확인 -----
print("\n[전처리 후] 데이터 크기:", df_jobs.shape)
print("제거된 행 수:", shape_before[0] - df_jobs.shape[0])

print("\n[전처리 후] 데이터 타입:")
display(df_jobs.dtypes)

print("\n[전처리 후] 결측값 개수:")
display(df_jobs.isna().sum())

print("\n[전처리 후] job_url 중복 개수:")
print(df_jobs.duplicated(subset=["job_url"]).sum())
````

### 실행 결과 해석 / 분석 / 요약

- 셀이 에러 없이 실행되었다.
- 전처리 전 데이터 크기는 5행 × 9열이고, 9개 컬럼이 모두 문자열(`str`) 타입이었다.
  - 즉 날짜처럼 보이는 `"2026-09-20"`도 pandas 입장에서는 그냥 글자였다.
- 날짜 컬럼 3개(`posted_date`, `closing_date`, `collected_at`)는 `datetime64[us]` 타입으로 변환되었다.
  - 이제 날짜 비교, 마감일까지 남은 기간 계산 같은 날짜 연산이 가능하다.
  - 변환 후 결측값이 0개이므로 `errors="coerce"` 때문에 `NaT`가 된 값은 없다. 즉 모든 날짜가 정상적으로 변환되었다.
- 문자열 컬럼 6개의 앞뒤 공백 제거를 적용했다. 샘플 데이터는 원래 공백이 없어서 값이 달라지지는 않았지만, 실제 수집 데이터에 대비한 기본 정리 절차를 갖추었다.
- `job_url` 기준 중복 제거 결과 제거된 행은 0개였고, 전처리 후 크기도 5행 × 9열로 그대로이다.
- 전처리 후 모든 컬럼의 결측값은 0개, `job_url` 중복 개수도 0개이다.
- 참고: pandas 3.x에서는 문자열 컬럼의 dtype이 `object` 대신 `str`로 표시된다.
- 결론: 전처리 / 중복 제거가 정상 완료되었으며, 다음 단계인 STEP 07 신규 공고 판별로 진행할 수 있다.

### STEP 07. 신규 공고 판별

### 작업 계획

STEP 06에서 전처리를 마친 `df_jobs`(공고 5건)를 이전 실행 이력과 비교해
**이번에 처음 나타난 공고(신규 공고)**만 골라낸다.

판별 기준:
- 1차 기준은 `job_url`이다. (URL이 같으면 같은 공고로 본다)
- 이전 이력에 있던 URL → 기존 공고
- 이전 이력에 없던 URL → 신규 공고

이전 실행 이력:
- 아직 실제 이력 파일이나 데이터베이스가 없으므로, 간단한 샘플 목록 `previous_urls`를 직접 만든다.
- 이전 이력에 `https://example.com/job/1`, `https://example.com/job/2`가 있었다고 가정한다.

확인할 항목:
- 이전 이력 URL 수
- 전체 공고 수 / 기존 공고 수 / 신규 공고 수
- 신규 공고의 `company_name`, `job_title`, `job_url`

이번 단계에서 하지 않는 것:
- STEP 08 기본 분석
- CSV 저장
- 실제 크롤링, Gemini API, Slack / Gmail

완료 조건:
- 기존 공고 수 + 신규 공고 수 = 전체 공고 수
- 신규 공고가 `new_jobs` DataFrame으로 만들어진다
- 신규 공고 목록에 이전 이력 URL이 들어 있지 않다

````python
# ----- 1) 이전 실행 이력(샘플) 만들기 -----
# 실제로는 이전 실행 때 저장해 둔 URL 목록을 불러와야 하지만,
# 지금은 연습을 위해 "이전에 이미 봤던 공고 URL" 2개를 직접 적어 둔다.
previous_urls = [
    "https://example.com/job/1",
    "https://example.com/job/2",
]
print("이전 이력 URL 수:", len(previous_urls))

# ----- 2) job_url이 이전 이력에 있는지 비교하기 -----
# isin(): 각 행의 job_url이 previous_urls 목록 안에 "들어 있는지" 한 번에 검사한다.
#         들어 있으면 True(기존 공고), 없으면 False(신규 공고)가 된다.
#         for 문으로 한 행씩 비교하지 않아도 되어서 간단하고 빠르다.
is_existing = df_jobs["job_url"].isin(previous_urls)

# ----- 3) 기존 공고 / 신규 공고 구분하기 -----
existing_jobs = df_jobs[is_existing]   # True인 행만 → 기존 공고
new_jobs = df_jobs[~is_existing]       # ~ 는 True/False를 뒤집는다 → 신규 공고

# ----- 4) 결과 출력 -----
print("\n전체 공고 수:", len(df_jobs))
print("기존 공고 수:", len(existing_jobs))
print("신규 공고 수:", len(new_jobs))

print("\n신규 공고 목록:")
display(new_jobs[["company_name", "job_title", "job_url"]])
````

### 실행 결과 해석 / 분석 / 요약

- 셀이 에러 없이 실행되었다.
- 이전 실행 이력(`previous_urls`)에는 URL 2개(`job/1`, `job/2`)가 있다.
- 현재 `df_jobs`의 전체 공고는 5건이다.
- `isin()`으로 비교한 결과:
  - 기존 공고(이전 이력에 있던 URL): 2건 → 회사A, 회사B
  - 신규 공고(이전 이력에 없던 URL): 3건 → 회사C, 회사D, 회사E
- 기존 2건 + 신규 3건 = 전체 5건이므로 빠지거나 겹친 공고가 없다.
- `new_jobs`에는 `job/3`, `job/4`, `job/5`만 들어 있고, 이전 이력 URL은 포함되지 않았다.
- 참고: 신규 공고의 행 번호(index)가 2, 3, 4로 남아 있는 것은 원래 `df_jobs`에서의 위치를 그대로 유지했기 때문이며 문제는 아니다.
- 한계: 지금의 `previous_urls`는 직접 만든 샘플이다. 실제 운영에서는 이번 실행의 URL을 저장해 두었다가 다음 실행 때 불러와야 한다. (저장은 이후 단계에서 진행)
- 결론: `job_url` 기준 신규 공고 판별이 정상 동작하며, 다음 단계인 STEP 08 기본 분석으로 진행할 수 있다.

처음부터 여러 페이지를 수집하지 않는 이유:
- 한 페이지에서 정확히 읽는 방법이 확인되지 않았는데 페이지를 늘리면, 틀린 데이터만 많아진다.

상태 코드 200만으로 성공 판단을 하면 안 되는 이유:
- STEP 03에서 상태 코드는 200이었지만, 실제로는 **보안정책(접속 제한) 페이지**가 돌아왔다.
- 그래서 상태 코드와 함께 Content-Type, 응답 길이, HTML `title`, "채용공고"/"AX" 문자열 포함 여부를 같이 확인한다.
- 기본 요청이 보안정책 페이지를 받으면, 일반 브라우저 User-Agent 헤더를 붙여 **딱 1회만** 다시 확인한다. (반복 재시도·우회 기법은 사용하지 않는다)

HTML 구조를 먼저 확인해야 하는 이유:
- 브라우저 화면과 HTML 코드 구조는 다를 수 있어서, 선택자를 추측하면 엉뚱한 값을 읽을 수 있다.
- 이번에는 실제 응답 HTML에서 공고 링크(`/Recruit/GI_Read/번호`)를 찾고, 그 부모 요소를 거슬러 올라가
  공고 1건을 감싸는 반복 블록(`div[data-sentry-component="CardJob"]`)을 먼저 확인한 뒤 선택자를 정했다.

날짜 데이터를 가져오는 방법:
- 화면 카드 HTML에는 등록일/마감일 **텍스트가 직접 들어 있지 않다.**
- 하지만 같은 HTML 응답 안의 `<script>`(Next.js가 심어 둔 `self.__next_f.push(...)` 데이터)에 공고별 JSON이 들어 있고, 여기에 날짜가 있다.
- 이 프로젝트에서는 **지원 제출 기간**이 중요하므로 다음 값을 사용한다.
  - 지원 제출 시작일 → `applicationPeriod.start`
  - 지원 제출 마감일 → `applicationPeriod.end`
- `createdAt`은 공고가 사이트에 **등록된 시각**이라 지원 기간과 다를 수 있으므로 사용하지 않는다.
  (예: 에스코어 공고는 `createdAt`=07-15, `applicationPeriod.start`=07-01)
- 카드 순서와 JSON 순서에 기대지 않고, **공고 ID**(`job_url` 끝의 숫자)로 카드와 JSON을 짝지어 날짜를 붙인다.

컬럼 의미 정의 (컬럼명은 STEP 02 명세를 유지):
- `posted_date` = 이번 프로젝트에서는 **지원 제출 시작일**을 의미한다. (`applicationPeriod.start`의 날짜 부분, 예: `2026-09-21`)
- `closing_date` = **지원 제출 마감일** (`applicationPeriod.end`의 날짜 + 시:분, 예: `2026-10-01 23:00`)
- 참고: 이름과 의미를 맞추기 위해, 이후 전체 리팩터링 단계에서 `application_start_date` / `application_end_date`로 바꿀 수 있다.
- `search_keyword` = `ax`, `collected_at` = 이번에 페이지를 받아 온 시각

추출 결과를 사람이 브라우저와 비교해야 하는 이유:
- 코드가 에러 없이 실행돼도 다른 칸의 글자를 읽었을 수 있다.
- 출력된 제목·회사명·지역·경력·날짜·URL을 브라우저 화면과 직접 비교해야 "제대로 읽었다"고 판단할 수 있다.

이번 단계에서 하지 않는 것:
- 여러 페이지 / 여러 검색어 / 상세 페이지 반복 요청
- DataFrame 생성(다음 STEP 07-B에서 확인), CSV·JSON 저장, 신규 공고 판별, 분석
- Gemini API, Slack / Gmail

성공 기준:
- 실제 검색 결과 HTML을 받았음을 확인한다
- 반복되는 공고 블록을 확인한다
- 최대 5건을 추출한다
- 최소 `company_name` 또는 `job_title`을 확인한다
- 가능하면 `job_url`도 확인한다
- `posted_date` / `closing_date`에 `applicationPeriod.start` / `end` 값이 들어간다
- 브라우저 화면과 비교할 수 있는 형태로 출력한다

````python
import re
import json
from datetime import datetime

import requests
from bs4 import BeautifulSoup

url = "https://www.jobkorea.co.kr/Search/?stext=ax&tabType=recruit"
search_keyword = "ax"
MAX_JOBS = 5  # 이번 STEP에서는 최대 5건만 읽는다


# 응답이 "실제 검색 결과"인지 확인하는 함수
# 상태 코드 200이어도 보안정책(접속 제한) 페이지일 수 있으므로 title과 본문 문자열까지 확인한다
def check_response(response):
    response.encoding = "utf-8"  # 한글이 깨지지 않도록 인코딩 지정
    soup = BeautifulSoup(response.text, "html.parser")
    title = soup.title.get_text(strip=True) if soup.title else None
    is_blocked = "보안정책" in response.text
    is_search_result = ("채용공고" in (title or "")) and ("AX" in response.text.upper())

    print("  상태 코드:", response.status_code)
    print("  Content-Type:", response.headers.get("Content-Type"))
    print("  응답 길이:", len(response.text))
    print("  HTML title:", title)
    print("  보안정책 페이지 여부:", is_blocked)
    print("  실제 검색 결과 여부:", is_search_result)
    return soup, is_search_result


# ----- 1) 가장 단순한 방식으로 1차 요청 -----
print("[1차 요청] requests.get() 기본 요청")
response = requests.get(url, timeout=10)
response.raise_for_status()
soup, is_search_result = check_response(response)

# ----- 2) 1차 요청이 보안정책 페이지일 때만, 일반 브라우저 User-Agent로 1회 추가 테스트 -----
used_user_agent = False
if not is_search_result:
    print("\n[2차 요청] 일반 브라우저 User-Agent 헤더를 붙여 1회만 다시 요청")
    headers = {
        "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 "
                      "(KHTML, like Gecko) Chrome/140.0.0.0 Safari/537.36"
    }
    response = requests.get(url, headers=headers, timeout=10)
    response.raise_for_status()
    soup, is_search_result = check_response(response)
    used_user_agent = True

# 수집 시각: 이번에 페이지를 받아 온 시각
collected_at = datetime.now().strftime("%Y-%m-%d %H:%M:%S")

# ----- 3) 실제 검색 결과가 아니면 여기서 중단 (재시도하지 않는다) -----
jobs = []
if not is_search_result:
    print("\n실제 검색 결과 HTML을 받지 못했습니다. 추출을 진행하지 않습니다.")
else:
    # ----- 4) Next.js script 안에서 공고 목록 JSON 찾기 -----
    # 화면 카드에는 날짜 텍스트가 없지만, 같은 HTML 안의 <script>에
    # self.__next_f.push([1, "...문자열..."]) 형태로 공고 데이터(JSON)가 들어 있다
    period_by_id = {}  # 공고 ID별 applicationPeriod 매핑: {"50032017": {"start": ..., "end": ...}}
    for script in soup.find_all("script"):
        text = script.string or ""
        if "applicationPeriod" not in text:
            continue
        match = re.match(r'self\.__next_f\.push\(\[1,(".*")\]\)', text, re.S)
        if not match:
            continue
        # 따옴표로 감싼 JavaScript 문자열 → 일반 문자열로 풀기
        payload = json.loads(match.group(1))
        # 문자열 안에서 공고 목록 "content": [ {...}, {...} ] 부분을 찾아 JSON으로 읽기
        for found in re.finditer(r'"content":\[\{"id":', payload):
            start_index = found.start() + len('"content":')
            content, _ = json.JSONDecoder().raw_decode(payload, start_index)
            # 공고 ID별 applicationPeriod 매핑 만들기
            for item in content:
                period_by_id[item["id"]] = item.get("applicationPeriod")
    print("\nJSON에서 찾은 공고 날짜 데이터 수:", len(period_by_id))

    # ----- 5) 반복되는 채용공고 블록 찾기 -----
    # 실제 HTML 확인 결과, 공고 1건은 data-sentry-component="CardJob" 인 div 하나로 감싸져 있었다
    cards = soup.select('div[data-sentry-component="CardJob"]')
    print("찾은 채용공고 후보(CardJob) 수:", len(cards))

    # ----- 6) 최대 5건만 필드 추출 -----
    for card in cards[:MAX_JOBS]:
        # 공고 제목 링크: data-sentry-component="Title" 인 <a>
        title_link = card.select_one('a[data-sentry-component="Title"]')
        # 회사명: 제목 링크 바로 다음에 나오는 <a>(같은 공고 URL) 안의 첫 번째 <span>
        company_link = title_link.find_next("a") if title_link else None
        company_tag = company_link.select_one("span") if company_link else None
        # 근무 지역: 첫 번째 GrayChip(📍 아이콘이 있는 칩) 안의 텍스트
        location_tag = card.select_one('[data-sentry-component="GrayChip"] span.truncate')
        # 경력 조건: "경력3년↑", "신입" 등이 들어 있는 작은 글씨 span
        career_tag = card.select_one("span.flex-shrink-0.text-typo-c1-13")

        # 공고 URL: 뒤에 붙은 ?Oem_Code=...&listno=... 는 검색 위치 추적용이라 제거하고 기본 주소만 남긴다
        job_url = title_link["href"].split("?")[0] if title_link else None

        # 화면 카드와 같은 ID의 날짜 붙이기 (순서가 아니라 공고 ID로 매칭한다)
        # 예: .../GI_Read/50032017 → job_id = "50032017"
        job_id = job_url.rsplit("/", 1)[-1] if job_url else None
        period = period_by_id.get(job_id) or {}
        start_raw = period.get("start")  # 예: "2026-09-21T00:00:00+09:00"
        end_raw = period.get("end")      # 예: "2026-10-01T23:00:00+09:00"

        jobs.append({
            "company_name": company_tag.get_text(strip=True) if company_tag else None,
            "job_title": title_link.get_text(strip=True) if title_link else None,
            "career": career_tag.get_text(strip=True) if career_tag else None,
            "location": location_tag.get_text(strip=True) if location_tag else None,
            # posted_date = 지원 제출 시작일 (applicationPeriod.start의 날짜 부분만)
            "posted_date": start_raw[:10] if start_raw else None,
            # closing_date = 지원 제출 마감일 (applicationPeriod.end의 날짜 + 시:분)
            "closing_date": end_raw[:16].replace("T", " ") if end_raw else None,
            "job_url": job_url,
            "search_keyword": search_keyword,
            "collected_at": collected_at,
        })

        # 원본 값도 함께 보여 주어 잘라 낸 결과와 비교할 수 있게 한다
        print(f"  ID {job_id} 원본 applicationPeriod: start={start_raw}, end={end_raw}")

# ----- 7) 결과 출력 (브라우저 화면과 비교용) -----
print("\nUser-Agent 추가 요청 여부:", used_user_agent)
print("실제 출력한 공고 수:", len(jobs))
for i, job in enumerate(jobs, start=1):
    print(f"\n[{i}]")
    for key, value in job.items():
        print(f"  {key}: {value}")
````

중복 기준: `job_url`
- 같은 회사가 공고를 여러 개 낼 수 있어서, 회사명만으로는 중복을 판단할 수 없다.

- `job_url`은 공고마다 번호가 달라서 개별 공고를 구분하는 데 가장 적합하다.
- STEP 07-A에서 `?Oem_Code=...&listno=...` 같은 추적용 값을 떼고 기본 공고 URL만 저장했으므로, 같은 공고는 같은 URL이 된다.

확인 내용:
- 현재 공고 수
- 중복된 `job_url`이 있는지
- 중복 행 수
- 중복 제거 후 공고 수
- 어떤 공고가 중복이었는지

중요:
- 원본 DataFrame인 `df_real_jobs`는 수정하지 않는다.
- 중복 제거 결과는 새로운 DataFrame `df_unique_jobs`에 저장한다.
- `df_real_jobs`가 메모리에 없으면 데이터를 임의로 만들지 않고, STEP 07-A → 07-B를 먼저 실행하라고 안내한다.

아직 하지 않는 것:

완료 조건:
- `job_url` 기준 중복 확인을 수행하고 중복 행 수를 확인한다
- `df_unique_jobs`가 만들어진다
- 원본 `df_real_jobs`가 그대로 보존된다
- 중복 제거 전/후 행 수를 확인한다

````python
# ----- 0) df_real_jobs가 있는지 먼저 확인 -----
# STEP 07-A → 07-B를 실행하지 않았다면 df_real_jobs가 없다 (데이터를 임의로 만들지 않는다)
if "df_real_jobs" not in globals():
    print("df_real_jobs가 없습니다. STEP 07-A와 STEP 07-B를 먼저 실행하세요.")
else:
    # ----- 1) 현재 행 수 -----
    print("현재 공고 수:", len(df_real_jobs))

    # ----- 2) job_url 기준 중복 여부 확인 -----
    # duplicated(): 중복인 행을 "찾기"만 한다 (True/False)
    # keep=False: 중복된 행을 첫 번째 것까지 모두 True로 표시 → 어떤 공고끼리 겹쳤는지 전부 볼 수 있다
    duplicate_mask = df_real_jobs.duplicated(subset=["job_url"], keep=False)
    duplicate_rows = df_real_jobs.loc[duplicate_mask, ["company_name", "job_title", "job_url"]]

    # ----- 3) 중복 행 수 -----
    print("중복 행 수:", duplicate_mask.sum())

    # ----- 4) 중복이 있으면 출력 -----
    if duplicate_mask.any():
        display(duplicate_rows)
    else:
        print("현재 5건 안에는 job_url 기준 중복 공고가 없습니다.")

    # ----- 5) 원본을 보존한 채 새 DataFrame 생성 -----
    # drop_duplicates(): 중복 행을 제거한 "새 결과"를 만든다 (df_real_jobs 자체는 바뀌지 않는다)
    # keep="first": 같은 job_url이 여러 개면 처음 나온 행만 남긴다
    # reset_index(drop=True): 행 번호를 0부터 다시 매긴다
    df_unique_jobs = (
        df_real_jobs
        .drop_duplicates(subset=["job_url"], keep="first")
        .reset_index(drop=True)
    )

    # ----- 6) 전/후 개수 확인 -----
    print("\n중복 제거 전:", len(df_real_jobs))
    print("중복 제거 후:", len(df_unique_jobs))
    print("제거된 행 수:", len(df_real_jobs) - len(df_unique_jobs))

    # ----- 7) 결과 확인 -----
    display(df_unique_jobs[["company_name", "job_title", "posted_date", "closing_date", "job_url"]])
````

````python
from pathlib import Path
import pandas as pd

# ----- 0) df_unique_jobs가 있는지 먼저 확인 -----
if "df_unique_jobs" not in globals():
    print("df_unique_jobs가 없습니다. STEP 07-A → 07-B → 07-C를 먼저 실행하세요.")
else:
    # ----- A) 경로 정의 -----
    # Notebook은 notebooks/ 폴더에서 실행되므로, 그 경우 한 단계 위를 프로젝트 루트로 사용한다
    current_dir = Path.cwd()
    project_root = current_dir.parent if current_dir.name == "notebooks" else current_dir
    history_path = project_root / "data" / "processed" / "jobs_history.csv"
    print("현재 실행 위치:", current_dir)
    print("history 파일 경로:", history_path)

    # ----- B) history 존재 여부 확인 -----
    if history_path.exists():
        history_df = pd.read_csv(history_path)
        print("\n기존 이력 파일 있음")
        print("기존 이력 건수:", len(history_df))
    else:
        # 과거 데이터가 없으므로 컬럼만 있는 빈 표로 시작한다
        history_df = pd.DataFrame(columns=df_unique_jobs.columns)
        print("\n기존 이력 파일 없음")
        print("첫 실행으로 처리합니다.")

    # ----- C) 신규 공고 판별 -----
    # set(): 이미 본 URL을 모아 둔 "집합" — 어떤 URL이 들어 있는지 빠르게 확인할 수 있다
    known_urls = set(history_df["job_url"].dropna().astype(str))

    # isin(known_urls): 현재 URL이 과거 URL 집합에 "있는지" True/False로 확인
    # ~ 를 붙이면 True/False가 뒤집힌다 → 과거 집합에 "없는" 공고만 선택 = 신규 공고
    new_jobs_df = df_unique_jobs[~df_unique_jobs["job_url"].isin(known_urls)].copy()

    print("\n현재 공고 수:", len(df_unique_jobs))
    print("기존 이력 수:", len(history_df))
    print("신규 공고 수:", len(new_jobs_df))

    # ----- E) 신규 공고 출력 -----
    if len(new_jobs_df) > 0:
        display(new_jobs_df[["company_name", "job_title", "posted_date", "closing_date", "job_url"]])
    else:
        print("이번 실행에서 새로 발견된 공고가 없습니다.")

    # ----- F) history 업데이트 -----
    # 과거 이력 + 현재 공고를 합친 뒤, job_url 기준으로 중복을 제거한다
    # keep="last": 같은 공고가 있으면 나중에 붙은 행(= 이번 실행의 최신 정보)을 남긴다
    updated_history_df = pd.concat([history_df, df_unique_jobs], ignore_index=True)
    updated_history_df = (
        updated_history_df
        .drop_duplicates(subset=["job_url"], keep="last")
        .reset_index(drop=True)
    )

    # 폴더가 없으면 만든다 (이미 있으면 그대로 사용)
    history_path.parent.mkdir(parents=True, exist_ok=True)
    # encoding="utf-8-sig": 엑셀에서 열어도 한글이 깨지지 않게 저장
    updated_history_df.to_csv(history_path, index=False, encoding="utf-8-sig")

    print("\n업데이트 후 이력 건수:", len(updated_history_df))
    print("history 저장 경로:", history_path)
````

DataFrame 역할 분리 (원본은 수정하지 않는다):
- `df_real_jobs` → 원본 실데이터
- `df_unique_jobs` → 중복 제거된 현재 공고 (새 컬럼도 추가하지 않는다)
- `df_filtered_jobs` → AX/AI 키워드 1차 필터 결과 (복사본이며, 여기에만 `matched_keywords` 컬럼을 추가한다)

이번 단계에서 하지 않는 것:
- Gemini API, 점수화, 추천 순위
- 상세 페이지 요청, 여러 페이지 / 여러 검색어 수집
- history CSV 갱신

완료 조건:
- 전체 공고 수, 필터 통과 공고 수, 제외 공고 수가 출력된다
- 공고별 회사명 / 제목 / 매칭 키워드를 사람이 확인할 수 있다
- `df_filtered_jobs`가 만들어지고, `df_unique_jobs`는 바뀌지 않는다

````python
# ----- 0) df_unique_jobs가 있는지 먼저 확인 -----
if "df_unique_jobs" not in globals():
    print("df_unique_jobs가 없습니다. 이전 실데이터 STEP을 먼저 실행하세요.")
else:
    # ----- 1) 1차 필터 키워드 -----
    relevance_keywords = [
        "AX",
        "AI",
        "인공지능",
        "생성형 AI",
        "생성형AI",
        "LLM",
        "Machine Learning",
        "머신러닝",
        "Data Scientist",
        "데이터 사이언티스트",
        "AI Engineer",
        "AI 엔지니어",
        "데이터 분석",
    ]

    # ----- 2) 제목에 키워드가 하나라도 있는지 확인하는 함수 -----
    def contains_relevance_keyword(title):
        if not isinstance(title, str):  # 제목이 비어 있으면(None 등) 통과시키지 않는다
            return False
        title_lower = title.lower()     # 비교할 때만 소문자로 바꾼다 (원본 값은 그대로)
        # any(): 여러 조건 중 하나라도 True면 True
        return any(keyword.lower() in title_lower for keyword in relevance_keywords)

    # ----- 3) 제목에서 실제로 매칭된 키워드 목록을 찾는 함수 -----
    def find_matched_keywords(title):
        if not isinstance(title, str):
            return []
        title_lower = title.lower()
        return [keyword for keyword in relevance_keywords if keyword.lower() in title_lower]

    # ----- 4) 1차 필터링 -----
    # apply(): job_title의 각 값에 함수를 하나씩 적용해서 True/False 목록을 만든다
    # .copy(): 필터 결과를 원본과 분리된 새 DataFrame으로 만든다
    is_relevant = df_unique_jobs["job_title"].apply(contains_relevance_keyword)
    df_filtered_jobs = df_unique_jobs[is_relevant].copy()

    # 매칭 키워드는 복사본(df_filtered_jobs)에만 컬럼으로 추가한다 (df_unique_jobs에는 추가하지 않음)
    df_filtered_jobs["matched_keywords"] = df_filtered_jobs["job_title"].apply(find_matched_keywords)

    # ----- 5) 결과 출력 -----
    print("전체 공고 수:", len(df_unique_jobs))
    print("1차 필터 통과 공고 수:", len(df_filtered_jobs))
    print("제외된 공고 수:", len(df_unique_jobs) - len(df_filtered_jobs))

    display(df_filtered_jobs[["company_name", "job_title", "career", "location", "posted_date", "closing_date"]])

    # ----- 6) 공고별 매칭 이유 확인 (사람이 직접 검토) -----
    print("\n=== 공고별 매칭 키워드 (전체 공고 기준) ===")
    for _, row in df_unique_jobs.iterrows():
        matched = find_matched_keywords(row["job_title"])
        result = "통과" if matched else "제외"
        print(f"[{result}] {row['company_name']} | {row['job_title']}")
        print(f"        → 매칭 키워드: {matched if matched else '없음'}")
````

# STEP 07-G. 키워드 필터 오탐 검증 및 개선

## 작업 계획

STEP 07-F의 필터는 `keyword in title` 방식이다.
이 방식은 **글자가 포함되어 있는지만** 보기 때문에, 다른 단어 속에 우연히 들어간 글자도 매칭할 수 있다. (오탐 = false positive)
- 예: `ai` → "M**ai**ntenance", "Tr**ai**ning" / `ax` → "T**ax**", "M**ax**"

이번 단계에서는 **실데이터를 다시 크롤링하지 않고**, 직접 만든 테스트 제목으로 이 문제를 확인하고 매칭 규칙을 개선한다.
(STEP 07-F의 함수는 수정하지 않고, 새 함수 `find_matched_keywords_safe()`를 만든다)

테스트 제목:

| 구분 | 제목 | 기대 결과 |
|---|---|---|
| 매칭되어야 함 | AX 컨설턴트 채용 | True |
| 매칭되어야 함 | AI Engineer 모집 | True |
| 매칭되어야 함 | LLM 개발자 | True |
| 매칭되어야 함 | 생성형 AI 서비스 개발 | True |
| 매칭되어야 함 | 머신러닝 엔지니어 | True |
| 매칭되어야 함 | 데이터 분석 담당자 | True |
| 매칭되면 안 됨 | Tax Manager | False |
| 매칭되면 안 됨 | Max Performance Engineer | False |
| 매칭되면 안 됨 | Maintenance Engineer | False |
| 매칭되면 안 됨 | Training Manager | False |
| 추가 (한글이 바로 붙은 경우) | AX담당자 모집 | True |
| 추가 (한글이 바로 붙은 경우) | AI기반 서비스 기획 | True |

개선 방식 — 키워드 종류에 따라 검사 방법을 나눈다:
- **영문 키워드** (`AX`, `AI`, `LLM`, `Machine Learning`, `Data Scientist`, `AI Engineer`)
  → 정규식으로 "키워드 **앞뒤에 영문자·숫자가 붙어 있지 않을 때만**" 매칭한다.
  - `Tax`: `ax` 앞에 영문자 `T`가 붙어 있으므로 매칭 안 됨
  - `AX 담당`, `(AX`, `AX담당`: 앞뒤가 공백·괄호·한글이므로 매칭됨
- **한글이 들어 있는 키워드** (`인공지능`, `생성형 AI`, `생성형AI`, `머신러닝`, `데이터 분석`, `AI 엔지니어`, `데이터 사이언티스트`)
  → 기존처럼 포함 검색을 사용한다.

왜 흔히 쓰는 `\b`(단어 경계)를 그대로 쓰지 않는가:
- Python 정규식은 **한글도 "단어 글자"로 취급**한다.
- 그래서 `\bAX\b`는 `"AX 담당"`은 잡지만, 한글이 바로 붙은 `"AX담당"`은 놓친다.
- 이런 이유로 모든 키워드에 같은 정규식을 쓰지 않고, 영문 키워드에만 "영문자·숫자 경계"를 적용한다.
- 코드에서 기존 방식, `\b` 방식, 개선 방식을 나란히 비교해 확인한다.

실데이터 적용:
- `df_unique_jobs`가 **현재 메모리에 있을 때만** 새 함수를 적용해 `df_filtered_jobs_safe`에 저장한다.
- 없으면 데이터를 임의로 만들지 않고 안내만 출력한다.
- 원본 `df_unique_jobs`, `df_filtered_jobs`는 수정하지 않는다.

주의:
- 정규식이 직무 의미를 완벽히 판단해 주지는 않는다. 이번 목표는 **명백한 오탐(Tax / Max / Maintenance / Training)을 줄이는 것**이다.
- 직무 의미의 최종 판단은 이후 Gemini 단계에서 한다.

이번 단계에서 하지 않는 것:
- 웹 재요청, 크롤링 재실행, history 갱신
- Gemini API, 상세 페이지 수집, 점수화

완료 조건:
- 테스트 12개가 모두 기대 결과와 일치한다
- 기존 방식에서 오탐이 생기는 제목을 출력으로 확인한다

````python
import re

# ----- 1) 키워드 (STEP 07-F와 같은 13개) -----
relevance_keywords = [
    "AX",
    "AI",
    "인공지능",
    "생성형 AI",
    "생성형AI",
    "LLM",
    "Machine Learning",
    "머신러닝",
    "Data Scientist",
    "데이터 사이언티스트",
    "AI Engineer",
    "AI 엔지니어",
    "데이터 분석",
]

# ----- 2) 테스트 데이터: (제목, 기대 결과) -----
test_titles = [
    ("AX 컨설턴트 채용", True),
    ("AI Engineer 모집", True),
    ("LLM 개발자", True),
    ("생성형 AI 서비스 개발", True),
    ("머신러닝 엔지니어", True),
    ("데이터 분석 담당자", True),
    ("Tax Manager", False),
    ("Max Performance Engineer", False),
    ("Maintenance Engineer", False),
    ("Training Manager", False),
    # 추가: 한글이 영문 키워드에 바로 붙은 경우 (\b 방식의 약점 확인용)
    ("AX담당자 모집", True),
    ("AI기반 서비스 기획", True),
]


# ----- 3) 기존 방식 (STEP 07-F와 같은 단순 포함 검색) — 비교용 -----
def find_matched_keywords_substring(title):
    title_lower = title.lower()
    return [k for k in relevance_keywords if k.lower() in title_lower]


# ----- 4) \b(단어 경계) 방식 — 비교용 -----
# Python에서는 한글도 "단어 글자"라서, "AX담당"처럼 한글이 바로 붙으면 경계로 인정되지 않는다
def find_matched_keywords_wordb(title):
    return [k for k in relevance_keywords
            if re.search(r"\b" + re.escape(k) + r"\b", title, re.IGNORECASE)]


# ----- 5) 개선 방식 -----
# 영문 키워드인지 확인: 알파벳·숫자·공백으로만 되어 있으면 영문 키워드로 본다
def is_english_keyword(keyword):
    return re.fullmatch(r"[A-Za-z0-9 ]+", keyword) is not None


def find_matched_keywords_safe(title):
    if not isinstance(title, str):
        return []
    matched = []
    for keyword in relevance_keywords:
        if is_english_keyword(keyword):
            # 영문 키워드: 앞뒤에 영문자·숫자가 "없을 때만" 매칭
            #   (?<![A-Za-z0-9]) → 바로 앞 글자가 영문자·숫자가 아니어야 함
            #   (?![A-Za-z0-9])  → 바로 뒤 글자가 영문자·숫자가 아니어야 함
            #   re.escape()      → 키워드 안의 특수문자를 글자 그대로 취급
            #   re.IGNORECASE    → 대소문자 구분 없이 비교
            pattern = r"(?<![A-Za-z0-9])" + re.escape(keyword) + r"(?![A-Za-z0-9])"
            if re.search(pattern, title, re.IGNORECASE):
                matched.append(keyword)
        else:
            # 한글이 들어 있는 키워드: 기존처럼 포함 검색
            if keyword.lower() in title.lower():
                matched.append(keyword)
    return matched


# ----- 6) 테스트 실행: 개선 방식이 기대 결과와 같은지 확인 -----
print("=== 개선 방식(find_matched_keywords_safe) 테스트 ===")
pass_count = 0
for title, expected in test_titles:
    matched = find_matched_keywords_safe(title)
    actual = len(matched) > 0
    result = "PASS" if actual == expected else "FAIL"
    if result == "PASS":
        pass_count += 1
    print(f"[{result}] {title}")
    print(f"    expected={expected}  actual={actual}  matched={matched}")
    if result == "FAIL":
        print("    → 실패 원인 확인 필요: 기대와 다른 키워드 매칭 결과")

print("\n전체 테스트:", len(test_titles))
print("통과:", pass_count)
print("실패:", len(test_titles) - pass_count)

# ----- 7) 세 방식 비교: 기존 방식에서 어떤 오탐/누락이 생기는지 확인 -----
print("\n=== 방식별 비교 (기대와 다르면 ✗) ===")
for title, expected in test_titles:
    row = []
    for name, func in [("기존 포함검색", find_matched_keywords_substring),
                       ("\\b 방식", find_matched_keywords_wordb),
                       ("개선 방식", find_matched_keywords_safe)]:
        matched = func(title)
        mark = "✓" if (len(matched) > 0) == expected else "✗"
        row.append(f"{name} {mark} {matched}")
    print(f"{title}  (기대={expected})")
    print("    " + " | ".join(row))

# ----- 8) 실데이터 적용 (df_unique_jobs가 메모리에 있을 때만) -----
print("\n=== 실데이터 적용 ===")
if "df_unique_jobs" not in globals():
    print("df_unique_jobs가 현재 메모리에 없어 실데이터 적용을 건너뜁니다.")
    print("실데이터에 적용하려면 STEP 07-A → 07-B → 07-C를 먼저 실행한 뒤 이 셀을 다시 실행하세요.")
else:
    safe_matches = df_unique_jobs["job_title"].apply(find_matched_keywords_safe)
    df_filtered_jobs_safe = df_unique_jobs[safe_matches.apply(len) > 0].copy()
    df_filtered_jobs_safe["matched_keywords"] = safe_matches[safe_matches.apply(len) > 0]

    print("전체 공고 수:", len(df_unique_jobs))
    print("개선 필터 통과 수:", len(df_filtered_jobs_safe))
    for title, company, matched in zip(df_unique_jobs["job_title"], df_unique_jobs["company_name"], safe_matches):
        print(f"[{'통과' if matched else '제외'}] {company} | {title}")
        print(f"        → 매칭 키워드: {matched if matched else '없음'}")
````

## 실행 결과 해석 / 분석 / 요약

# STEP 07-I. Gemini API 최소 연결 테스트

## 작업 계획

STEP 07-H에서는 `GEMINI_API_KEY`가 **있는지**만 확인했다.
이번 단계에서는 그 키로 **실제로 Gemini API를 1번 호출**해서, 연결이 정상인지 확인한다.

이번 단계 목적:
- `GEMINI_API_KEY`가 실제 API 호출에 사용 가능한지 확인
- Gemini 서버와 통신이 되는지 확인
- 응답 텍스트를 정상적으로 받을 수 있는지 확인

사용 도구:
- Google 공식 Python SDK **`google-genai`** (`from google import genai`)
  - 예전 패키지인 `google-generativeai`는 사용하지 않는다.
- 모델: `gemini-3.6-flash` (Flash 계열)
  - 이 모델을 쓸 수 없다는 오류가 나오면, 다른 모델을 여러 번 시도하지 않고 오류를 기록한 뒤 사용 가능한 모델을 확인한다.

보내는 내용:
- 아주 짧은 연결 테스트 문장 **1개만** 보낸다. (비용과 호출 횟수를 최소화)

보안 원칙:
- API 키 값, 키 일부, 키 길이, `.env` 내용은 **절대 출력하지 않는다.**
- 오류가 나면 오류 종류와 메시지만 출력하되, 메시지 안에 키가 섞여 있을 경우를 대비해 키 부분은 `***`로 가린다.

아직 하지 않는 것:
- 채용공고 5건 요약
- 반복 호출
- JSON 구조화
- 비용이 커질 수 있는 긴 프롬프트
- 상세 페이지 내용 전달
- Slack / Gmail 연동

성공 기준:
- API 호출 오류 없음
- Gemini가 짧은 텍스트 응답을 반환
- API 키가 화면에 출력되지 않음

````python
from pathlib import Path
import os
from dotenv import load_dotenv
from google import genai

# ----- 1) 프로젝트 루트의 .env 읽기 (STEP 07-H와 같은 방식) -----
current_dir = Path.cwd()
project_root = current_dir.parent if current_dir.name == "notebooks" else current_dir
env_path = project_root / ".env"

print(".env 파일 존재 여부:", env_path.exists())
if env_path.exists():
    load_dotenv(env_path)

# ----- 2) API 키 확인 (값은 절대 출력하지 않는다) -----
gemini_api_key = os.getenv("GEMINI_API_KEY")
print("GEMINI_API_KEY 설정 여부:", "설정됨" if gemini_api_key else "미설정")

MODEL_NAME = "gemini-3.6-flash"

if not gemini_api_key:
    print("\nGEMINI_API_KEY가 없어 API를 호출하지 않습니다.")
    print("프로젝트 루트의 .env 파일에 GEMINI_API_KEY=본인의_API_키 를 입력한 뒤 다시 실행하세요.")
else:
    # ----- 3) Gemini Client 생성 -----
    # client: Gemini API에 요청을 보내기 위한 연결 객체
    client = genai.Client(api_key=gemini_api_key)

    # ----- 4) 아주 짧은 프롬프트 1개만 전송 -----
    test_prompt = (
        "연결 테스트입니다. "
        "한국어로 'Gemini API 연결 성공'이라고 짧게 답해 주세요."
    )
    print("\n사용 모델:", MODEL_NAME)

    try:
        # generate_content(): 프롬프트를 보내고 응답을 받는 함수 (이 셀에서 1번만 호출)
        response = client.models.generate_content(
            model=MODEL_NAME,
            contents=test_prompt,
        )
        # response.text: Gemini가 만든 텍스트 결과
        print("\nGemini 응답:")
        print(response.text)
    except Exception as e:
        # 오류 메시지 안에 혹시 키가 들어 있으면 *** 로 가린 뒤 출력한다
        safe_message = str(e).replace(gemini_api_key, "***")
        print("\nGemini API 호출 실패")
        print("오류 종류:", type(e).__name__)
        print("오류 메시지:", safe_message[:500])
````

## 실행 결과 해석 / 분석 / 요약

- .env 로드 여부: **로드됨** — 프로젝트 루트의 `.env`를 찾아 `load_dotenv()`로 읽었다.
- GEMINI_API_KEY 설정 여부: **설정됨**
- 사용한 모델: `gemini-3.6-flash`
- API 호출 성공 여부: **성공**

- 오류가 있었다면 오류 종류: 오류 없음.
  - 참고: 출력 맨 위의 `Direct use of automatic function calling (AFC) ...` 문구는 오류가 아니라 SDK가 보여 주는 **안내(경고) 메시지**이다.
    AFC는 Gemini가 Python 함수를 대신 호출하게 하는 기능인데, 이번 호출에서는 사용하지 않았으므로 결과에 영향이 없다.
- 다음 단계 진행 가능 여부: **진행 가능** — `.env → 키 로드 → Client 생성 → 짧은 프롬프트 1회 → response.text 수신` 흐름이 실제로 동작했다.

주의:
- 이 셀을 실행할 때마다 API가 1번 호출된다. 불필요하게 반복 실행하지 않는다.
- 채용공고 요약처럼 긴 내용을 보내는 작업은 다음 단계에서 호출 횟수와 입력 길이를 정해 진행한다.

#### 초보자 설명
- `client` → Gemini API에 요청을 보내기 위한 **연결 객체**이다. API 키를 넣어 한 번 만들어 두고 계속 사용한다.
- `generate_content()` → 프롬프트를 보내고 **응답을 받는 함수**이다. `model`에 사용할 모델, `contents`에 보낼 내용을 넣는다.
- `response.text` → Gemini가 생성한 **텍스트 결과**이다.
